```python
# Remove columns with too many missing values or unnecessary information for our Chatbot.
# We discussed that total amount of reviews are more relevant than monthly count and last review dates.
clean_df = df.drop(columns=["neighbourhood_group", "latitude", "longitude", "last_review", "reviews_per_month", "scrape_date"])
```

```python
# A box plot to show the distribution of prices and the outliers.
# Too many outliers and price range is too wide to show the distribution of prices for the majority of listings.
sns.boxplot(x=clean_df["price"])
```

```python
# Check the distribution of prices for different room types to decide if we should remove outliers.
clean_df.groupby("room_type")["price"].describe()
```

```python
# Remove listings in non european cities, so our chatbot can focus on countries with similar context.
clean_df = clean_df[(clean_df["city"] != "Bangkok") & (clean_df["city"] != "Sydney")]
```

```python
# 444 listings with price lower than 25 USD per night
# Cheapest we could find on the Airbnb official website was about 30 USD per night
# Those extremely cheap listings did exist on different platforms, so we decided to keep them.

clean_df[clean_df["price"] < 25][["name", "city", "room_type", "price", "minimum_nights"]].sort_values("price")
```

```python
# Check the information of the extreamly expensive listings again, after removing the non european cities.
clean_df[clean_df["price"] > 3000][["name", "city", "room_type", "price", "minimum_nights"]].sort_values("price", ascending=False)
```

```python
# Show the top 0.5% of the prices to decide on the limit for the outliers.
clean_df["price"].quantile(0.995)
```

```python
# Remove listings with price higher than 1800 USD per night, which is the top 0.5% of the prices.
clean_df = clean_df[clean_df["price"] <= 1800]
```

```python
# We can see that this column has many types of values, and it needs to be handled.
clean_df["license"].value_counts()
```

```python
# Many values in license column are missing, so we need to make a category for it.
clean_df["license"].isna().sum()
```

```python
# Rename few values to "Without license" and other names for better understanding
license_mapping = {
    "Exempt": "Without license",
    "I applied for a CIN, but haven’t received it yet": "Without license",
}

# Replace certain values to the categories we created
clean_df["license"] = clean_df["license"].replace(license_mapping) 

# Missing values become Unknown
clean_df["license"] = clean_df["license"].fillna("Unknown")

# Everything that is not in the license_mapping values becomes "With license" (This part can be deeper investigated for future work)
clean_df["license"] = clean_df["license"].where(clean_df["license"].isin(["Without license", "Unknown"]),"With license")
```

```python
# Unknown is the biggest category group, discuss later if we want to drop it or not.
clean_df["license"].value_counts()
```

```python
# See how many values exist for each threshold in minimum_nights
print((clean_df["minimum_nights"] > 90).sum())
print((clean_df["minimum_nights"] > 180).sum())
print((clean_df["minimum_nights"] > 365).sum())
```

```python
# Remove rows where minimum_nights is above 90, because travel visa is usually valid for 90 days and it is not relevant for our Chatbot to have listings with minimum_nights above 90.
clean_df = clean_df[clean_df["minimum_nights"] <= 90]
clean_df.head(50)
```

```python
# Check how many rows exist now
clean_df.shape
```

```python
# Here we try to reduce the size of the dataset.
# Drop all the rows where price is null
clean_df = clean_df.dropna(subset=["price"])

# Drop host_name columns
clean_df = clean_df.drop(columns=["host_name"])

# Drop all the rows where number_of_reviews is 0 to only include more trustworthy listings
clean_df = clean_df[clean_df['number_of_reviews'] > 0]

# Drop all the rows where license is Without license for the same reason as above
clean_df = clean_df[clean_df['license'] != 'Without license']

clean_df.head(50)
```
